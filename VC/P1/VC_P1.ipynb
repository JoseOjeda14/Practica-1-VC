**Código Realizado por la IA**

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Crear el lienzo completamente negro (0) de 800x800 píxeles
tablero = np.zeros((800, 800, 1), dtype=np.uint8)

# 2. Definir el tamaño de cada casilla (800 píxeles / 8 casillas = 100 píxeles)
tamano_casilla = 100

# 3. Lógica matemática para alternar los colores
for fila in range(8):
    for columna in range(8):
        # En el ajedrez, si la suma del índice de la fila y la columna es par, es de un color
        if (fila + columna) % 2 == 0:
            
            # Calculamos las coordenadas exactas de la casilla en los ejes Y y X
            y_inicio = fila * tamano_casilla
            y_fin = (fila + 1) * tamano_casilla
            x_inicio = columna * tamano_casilla
            x_fin = (columna + 1) * tamano_casilla
            
            # Pintamos ese bloque específico de blanco absoluto (255)
            tablero[y_inicio:y_fin, x_inicio:x_fin, 0] = 255

# 4. Visualizar la textura resultante
plt.imshow(tablero, cmap='gray', vmin=0, vmax=255)
plt.title('Textura Tablero de Ajedrez 800x800 (Generado por IA)')
plt.axis('off') # Ocultamos los números de los ejes para ver solo la textura
plt.show()
```

**TAREA 4**: Llevar a cabo una propuesta propia de pop art.

```python
vid = cv2.VideoCapture(0)

# Dimensiones de la cámara
w = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Fuerzo a mitad de resolución para ocupar menos pantalla
w = int(w/2)
h = int(h/2)
vid.set(cv2.CAP_PROP_FRAME_WIDTH, w)
vid.set(cv2.CAP_PROP_FRAME_HEIGHT, h)

# Imagen conjunta 2x original
collage = np.zeros((h*2, w*2, 3), dtype = np.uint8)
tl = collage[0:h, 0:w]
tr = collage[0:h, w:w+w]
bl = collage[h:h+h, 0:w]
br = collage[h:h+h, w:w+w]

while True:
    # fotograma a fotograma
    ret, frameIN = vid.read()
    if not ret:
        break

    # Menor tamaño
    frame = cv2.resize(frameIN, (int(w), int(h)), cv2.INTER_NEAREST)

    # Estilo Roy Lichtenstein

    # Posterizado (4 tonos fijos)
    color_plano = (frame // 64) * 64 + 32
    
    # Delineado de bordes
    gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gris_suav = cv2.medianBlur(gris, 7)
    bordes = cv2.adaptiveThreshold(gris_suav, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 9, 2)
    
    # Juntamos colores planos y bordes negros
    comic = cv2.bitwise_and(color_plano, color_plano, mask=bordes)

    # Separamos canales
    r = comic[:,:,2]
    g = comic[:,:,1]
    b = comic[:,:,0]

    # Jugamos con los valores de los planos
    tl[:,:,0] = b
    tl[:,:,1] = g
    tl[:,:,2] = r

    tr[:,:,0] = 255 - r
    tr[:,:,1] = g
    tr[:,:,2] = b
    
    bl[:,:,0] = r
    bl[:,:,1] = 255 - b
    bl[:,:,2] = g

    br[:,:,0] = b
    br[:,:,1] = g
    br[:,:,2] = 255 - r

    # Muestra composición
    cv2.imshow('Cam', collage)

    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break

# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```
